矩阵与向量相乘的例子

```python
import numpy as np  
  
# 定义矩阵A和列向量v  
A = np.array([[1, 2, 3],  
              [4, 5, 6]])  
  
v = np.array([[7],  
              [8],  
              [9]])  
  
# 注意：由于v是一个列向量，我们需要确保它的形状是(3, 1)而不是(3,)  
# 如果v的形状是(3,)，则可以通过v.reshape(-1, 1)将其转换为列向量  
# 但在这个例子中，我们直接定义了v为列向量形状  
  
# 矩阵与向量相乘  
result = np.dot(A, v)  
  
# 打印结果  
print("矩阵A:")  
print(A)  
print("列向量v:")  
print(v)  
print("矩阵A与列向量v相乘的结果:")  
print(result)
```

向量与矩阵相乘的例子

```python
import numpy as np  
  
# 定义行向量u（注意：这里我们可以将其定义为二维数组的形式）  
#u = np.array([[1, 2, 3]]) 

#定义为一维数组
u = np.array([1, 2, 3])  


  
# 定义矩阵B  
B = np.array([[4, 5],  
              [6, 7],  
              [8, 9]])  
  
# 行向量u与矩阵B相乘（注意：这里不需要转置u，因为u已经是一个行向量）  
result = np.dot(u, B)  
  
# 打印结果  
print("行向量u:")  
print(u)  
print("矩阵B:")  
print(B)  
print("行向量u与矩阵B相乘的结果:")  
print(result)
```

## 3.1 Abstract
<img align="right" src="./images/C1_W2_Lab04_Vectors.PNG" style="width:340px;" >

在本课程中，你将要使用的向量是数字的有序数组。在符号表示中，向量用小写粗体字母表示，如 $\mathbf{x}$。向量的所有元素都是同一类型的。例如，向量不会同时包含字符和数字。向量中元素的数量通常被称为“维度(dimension)”，尽管数学家可能更喜欢使用“秩(rank)”这个词。右图所展示的向量具有n维。向量的元素可以通过索引来引用。在数学环境中，索引通常从 1 到 n。在计算机科学和这些实验室中，索引通常从 0 到 n-1。在符号表示中，当单独引用向量的元素时，会在下标中指示索引，例如，向量 $\mathbf{x}$ 的第0个元素是$x_0$。请注意，在这种情况下，$x$ 不是粗体的。

Vectors, as you will use them in this course, are ordered arrays of numbers. In notation, vectors are denoted with lower case bold letters such as $\mathbf{x}$.  The elements of a vector are all the same type. A vector does not, for example, contain both characters and numbers. The number of elements in the array is often referred to as the *dimension* though mathematicians may prefer *rank*. The vector shown has a dimension of $n$. The elements of a vector can be referenced with an index. In math settings, indexes typically run from 1 to n. In computer science and these labs, indexing will typically run from 0 to n-1.  In notation, elements of a vector, when referenced individually will indicate the index in a subscript, for example, the $0^{th}$ element, of the vector $\mathbf{x}$ is $x_0$. Note, the $x$ is not bold in this case.

## 3.2 NumPy Arrays

NumPy的基本数据结构是一个可索引的、$n$维的数组，其中包含相同类型（dtype）的元素。你可能立刻就注意到了我们对“维度”这个词进行了重载。在上面的描述中，它指的是向量中元素的数量，而在这里，维度指的是数组索引的数量。一维或1-D数组有一个索引。在第一门课程中，我们将使用NumPy 1-D数组来表示向量。（要会区分概念上的差异：2维数组和2维向量在语义层面的差异，参见下面的PS）

NumPy's basic data structure is an indexable, n-dimensional *array* containing elements of the same type (`dtype`). Right away, you may notice we have overloaded the term 'dimension'. Above, it was the number of elements in the vector, here, dimension refers to the number of indexes of an array. A one-dimensional or 1-D array has one index. In Course 1, we will represent vectors as NumPy 1-D arrays. 

 - 一维数组，形状（n,）：有n个元素，索引从[0]到[n-1]
 - 1-D array, shape (n,): n elements indexed [0] through [n-1]
 
 
 ###  PS 
   如果你使用 numpy, pytorch, tensorflow，给定任意一个权重矩阵$\mathbf{W}$，一般呈现以下规律：
 - 形状(n,)表示$n$维向量或者具有$n$个元素的一维数组
 - 形状(n, m)表示二维数组或矩阵，维度是 $\mathbf{W} \in {\mathbb{R}^{{ n \times m}}}$          
 - 形状(n, m, k)表示三维数组或矩阵，维度是 $\mathbf{W} \in {\mathbb{R}^{{ n \times m \times k}}}$   
 - 形状(n, m, k, z)表示四维数组或矩阵，维度是 $\mathbf{W} \in {\mathbb{R}^{{ n \times m \times k \times z}}}$ 
 - 这里举例所使用的字母没有什么特殊含义，仅用于理解数组的形状。

Data creation routines in NumPy will generally have a first parameter which is the shape of the object. This can either be a single value for a 1-D result or a tuple (n,m,...) specifying the shape of the result. Below are examples of creating vectors using these routines.

### 3.4.1 Indexing
可以通过索引和切片访问向量的元素。NumPy提供了一套非常完整的索引和切片功能。在这里，我们仅探讨本课程所需的基础知识。有关更多详细信息，请参考[Slicing and Indexing](https://numpy.org/doc/stable/user/basics.indexing.html#) for more details.  

中文连接https://www.numpy.org.cn/reference/arrays/indexing.html

Elements of vectors can be accessed via indexing and slicing. NumPy provides a very complete set of indexing and slicing capabilities. We will explore only the basics needed for the course here. Reference [Slicing and Indexing](https://numpy.org/doc/stable/user/basics.indexing.html#) for more details.  

**索引**是指通过数组中的位置来引用数组的一个元素。
**Indexing** means referring to *an element* of an array by its position within the array.

**切片**是指根据元素的索引从数组中获取一个子集。
**Slicing** means getting a *subset* of elements from an array based on their indices.  

NumPy的索引从0开始，因此向量$\mathbf{a}$的第三个元素是`a[2]`.
NumPy starts indexing at zero so the 3rd element of an vector $\mathbf{a}$ is `a[2]`.



```python
#vector indexing operations on 1-D vectors
a = np.arange(10)
print(a)

#access an element
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# access the last element, negative indexes count from the end
print(f"a[-1] = {a[-1]}")

#indexs must be within the range of the vector or they will produce and error
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

(The dot product is a mainstay of Linear Algebra and NumPy. This is an operation used extensively in this course and should be well understood. The dot product is shown below.)

<img src="./images/C1_W2_Lab04_dot_notrans.gif" width=800> 

点积将两个向量中对应位置的元素相乘，然后将结果相加。(The dot product multiplies the values in two vectors element-wise and then sums the result.)

向量点积要求两个向量的维度必须相同。(Vector dot product requires the dimensions of the two vectors to be the same.)

下面我们来实现自己实现点积操作：Let's implement our own version of the dot product below:

**使用for循环**，实现一个函数，该函数返回两个向量的点积。给定输入 $a$ 和 $b$，函数返回：

**Using a for loop**, implement a function which returns the dot product of two vectors. The function to return given inputs $a$ and $b$:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$

假设`a`和`b`的形状相同。
Assume both `a` and `b` are the same shape.

```python
def my_dot(a, b): 
    """
   Compute the dot product of two vectors
 
    Args:
      a (ndarray (n,)):  input vector 
      b (ndarray (n,)):  input vector with same dimension as a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
```

```python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

Note, the dot product is expected to return a scalar value(返回一个标量值). 

Let's try the same operations using `np.dot`.  

```python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
```

### 3.4.7 The Need for Speed: vector vs for loop (对速度的需求：向量 vs for循环)

我们使用NumPy库是因为它提高了计算速度。让我们来演示一下：

We utilized the NumPy  library because it improves speed memory efficiency. Let's demonstrate:


```python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
```
