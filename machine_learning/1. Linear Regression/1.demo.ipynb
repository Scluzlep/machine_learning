```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_regression#https://scikit-learn.org.cn/view/595.html
from sklearn import linear_model
#plt.style.use('./deeplearning.mplstyle')
```

# 生成随机数据

```python
X, y, w = make_regression(n_samples=100, n_features=1, coef=True, noise=20.0,#调整噪声值
                          random_state=1, bias=3.5)#n_samples 样本数
```

```python
print("X.shape",X.shape)
print("y.shape",y.shape)
print("w: ",w)#线性模型的权重系数
```

```python
plt.scatter(X, y, marker='x', c='r')
```

```python
lr = linear_model.LinearRegression()
lr.fit(X, y)
print("lr.coef_",lr.coef_)#权重系数 w
print("lr.intercept_",lr.intercept_)#截距 即偏差项
```

```python
np.arange(X.min(), X.max()).shape
```

```python
line_X = np.arange(X.min(), X.max())[:, np.newaxis]# np.newaxis 在数组的指定位置增加一个新的轴（维度）
line_y = lr.predict(line_X)
```

```python
plt.scatter(X, y, marker='x', c='r')
plt.plot(line_X, line_y, color='navy', label='Linear regressor')
plt.legend()
plt.show
```
